```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('train.txt',sep = ';',header = None, names = ['text','emotion'])
#csv file is a coma(',') separated file , here we are use ';' for separation and use txt file as csv
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
uni_emotions = df['emotion'].unique()
emotion_numbers = {}
i=0
for emo in uni_emotions:
  emotion_numbers[emo] = i
  i += 1

df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
df
```

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
df
```

```python
import string

def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

```python
df
```

```python
def remove_numbers(txt):
  new = ""
  for i in txt:
    if not i.isdigit():
      new += i
  return new

df['text'] = df['text'].apply(remove_numbers)
```

```python
df
```

```python
def remove_emojis(txt):
  # ASCII values are alwyas assigned with (A,B....a,b)
  # uvicorn are always assign with emojis

  new =""
  for i in txt :
    if i.isascii():
      new += i
  return new

df['text'] = df['text'].apply(remove_emojis)
```

```python
df
```

```python
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')

```

```python
stop_words = set(stopwords.words('english'))
```

```python
df.loc[1]['text']
```

```python
df
```

```python
def remove(txt):
  words = txt.split()
  cleaned = []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)
  return ' '.join(cleaned)

df['text'] = df['text'].apply(remove)
df.loc[1]['text']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(df['text'],df['emotion'],test_size = 0.2,random_state = 42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer

bow_vectorizer = TfidfVectorizer()
bow_X_train = bow_vectorizer.fit_transform(X_train)
bow_X_test = bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

model = MultinomialNB()
model.fit(bow_X_train,y_train)
```

```python
pred_bow = model.predict(bow_X_test)
print(accuracy_score(y_test, pred_bow))
```

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_X_train = tfidf_vectorizer.fit_transform(X_train)
tfidf_X_test = tfidf_vectorizer.transform(X_test)

nb2_model = MultinomialNB()
nb2_model.fit(tfidf_X_train,y_train)
```

```python
y_pred = nb2_model.predict(tfidf_X_test)
print(accuracy_score(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression


logistic_model = LogisticRegression(max_iter=1000)
logistic_model.fit(tfidf_X_train, y_train)

log_pred = logistic_model.predict(tfidf_X_test)
print(accuracy_score(y_test, log_pred))
```
